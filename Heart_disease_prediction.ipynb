```python

# STEP 1 : Import Libraries
# ==============================

import pandas as pd

# pandas is used for:
# -> Data handling
# -> Data cleaning
# -> Working with tables and datasets

import numpy as np

# numpy is used for:
# -> Numerical calculations
# -> Mathematical operations
# -> Array handling

import matplotlib.pyplot as plt

# matplotlib is used for:
# -> Data visualization
# -> Creating graphs and charts

from sklearn.model_selection import train_test_split

# train_test_split is used for:
# -> Dividing dataset into training and testing data
# -> Training data teaches model
# -> Testing data checks performance

from sklearn.linear_model import LogisticRegression

# LogisticRegression is a classification algorithm
# It predicts categories like:
# -> Yes / No
# -> True / False
# -> 0 / 1
# Example:
# -> Customer Purchase Prediction
# -> Spam Email Detection
# -> Disease Prediction

from sklearn.metrics import accuracy_score

# accuracy_score checks:
# -> How many predictions are correct
# Higher accuracy means better model

from sklearn.metrics import confusion_matrix

# confusion_matrix shows:
# -> Correct predictions
# -> Wrong predictions
# -> Model performance in table form

from sklearn.metrics import classification_report

# classification_report gives:
# -> Precision
# -> Recall
# -> F1-Score
# -> Overall classification performance

```

STEP#2

```python
import pandas as pd

# Creating dictionary
data = {
    "Patient_ID": [300001, 300002, 300003, 300004, 300005],
    "Age": [57, 35, 68, 57, 64],
    "Gender": [1, 1, 0, 0, 1],
    "Chest_Pain_Type": [0, 0, 0, 2, 3],
    "Resting_BP": [161, 128, 157, 115, 100],
    "Cholesterol": [341, 234, 123, 271, 358],
    "Fasting_BS": [1, 0, 0, 0, 1],
    "Max_Heart_Rate": [185, 93, 85, 145, 150],
    "Exercise_Angina": [1, 0, 1, 0, 1],
    "Oldpeak": [3.0, 3.2, 0.2, 3.0, 4.5],
    "Smoking": [1, 1, 1, 1, 0],
    "Heart_Disease": [1, 0, 0, 1, 1]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Display dataframe
print(df)
```

Step 3 # Understand the data

```python
print("\nDATASET SHAPE")

# shape gives:
# -> Total rows
# -> Total columns

print(df.shape)

print("\nCOLUMN NAMES")

# columns displays all column names

print(df.columns)

print("\nDATA TYPES")

# dtypes checks datatype of columns

print(df.dtypes)

print("\nMISSING VALUES")

# isnull() checks missing values
# sum() counts missing values



print(df.isnull().sum())# it  count  all  the  null  value in  a  particular  columns
print(df.isnull().sum().sum())# Count  all  the  null value  in  the  datset.

print("\nDuplicate values")
print(df.duplicated().sum()) # it count all the duplicates value
```

step 4 # Label Encoding

```python
# Import labelencoder

from sklearn.preprocessing import LabelEncoder

categorical_col=df.select_dtypes(include=['object']).columns
print(categorical_col)

le=LabelEncoder()
for col in categorical_col:
  df[col]=le.fit_transform(df[col])

print(df.head())
```

STEP 4 : DEFINE FEATURES AND TARGET

```python
X = df.drop(columns=['Heart_Disease'],axis=1)

# X contains input feature


# Model learns using this feature

y = df['Heart_Disease']

# y contains target variable

#  Heart_Disease  is output variable
# Model predicts this value
```

STEP 5 : SPLIT DATASET

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.2,

    # 20% data used for testing

    random_state=42

    # Same output every execution
)

print("\nTRAINING DATA SIZE :", len(X_train))

# Number of training rows

print("TESTING DATA SIZE :", len(X_test))

# Number of testing rows
```

Step 6 # Create LinearRegreesion Model

```python
model=LogisticRegression()
print("Model initiated Successfully")
```

Step#7 Train Model

```python
model.fit(X_train,y_train)
print("Model Trained Successfully")
```

step #8 Make Prediction

```python
y_pred=model.predict(X_test)

# Display the first 10 actual values

print("\nActual Values")
print(y_test.values[:10])

# display the first 10 predicted values

print("\nPredicted Values")
print(y_pred[:10])
```

Step #9  Model Evaluation

```python
accuracy=accuracy_score(y_test,y_pred)

# accuracy_score checks:
# -> Total correct predictions
# Formula:
# Correct Predictions / Total Predictions
# Higher accuracy means better model

cm=confusion_matrix(y_test,y_pred)

# confusion_matrix creates performance table
# It shows:
# -> True Positive
# -> True Negative
# -> False Positive
# -> False Negative
# Helps understand prediction mistakes


report=classification_report(y_test,y_pred)

# classification_report gives:
# -> Precision
# -> Recall
# -> F1-Score
# -> Support
# Used for detailed model evaluation

print("\nAccuracy Score")
print(accuracy)

print("\nConfusion Matrix")
print(cm)

print("\nClassification Report")
print(report)
```

Step 10 # CHECK MODEL COEFFICIENTS

```python
print("model coefficients")
# coef_ gives slope value
print(model.coef_)

print("model intercept")
# intercept_ gives starting/base value
print(model.intercept_)
```
